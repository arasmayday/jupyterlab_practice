```python
def compute_mean(values):
    mean = sum(values) / len(values)
    print(mean)

compute_mean([2, 4, 6, 8])
```

The function compute_mean calculates the mean of the values in a list. 
It adds all the values and divides the total by the number of values in the list.
The mean can be expressed in the following Latex:
$$
\bar{x} = \frac{x_1+x_2+\cdots+x_n}{n}
$$

and 

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$
